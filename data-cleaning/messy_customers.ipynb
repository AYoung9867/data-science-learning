```python
#Remove duplicate column
df = df.drop_duplicates()
print(df[df.duplicated()])
```

```python
#change negative age to NaN
df.loc[df['age']<0, 'age'] = np.nan
df.describe()
```

```python
#store median salary from full dataset
salary = df['salary'].median()

#drop rows with Null age or date
df = df.dropna(subset=['age', 'signup_date'])

#Replace Nan with median salary
df.loc[df['salary'].isna(), 'salary'] = salary

#Check all NaNs are removed
df.isna().sum()
```

```python
#Solve LA problem
df.loc[df['city'] == 'LA', 'city'] = 'Los Angeles'

#remove surrounding whitespace and case standardisation
df['city'] = df['city'].str.strip().str.title()

df['city'].unique()
```

```python
#remove surrounding whitespace and case standardisation
df['name'] = df['name'].str.strip().str.title()

df['name'].unique()
```

```python
#sort Dtype of date
df['signup_date'] = pd.to_datetime(df['signup_date'])

df.info()
```
